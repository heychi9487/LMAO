# Decision tree regression

```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
import matplotlib.pyplot as plt
%matplotlib inline

dataset = pd.read_csv("slr-salary.csv")
X = np.array(dataset["lv"]).reshape(-1, 1)
y = dataset["salary"]

reg1 = DecisionTreeRegressor()

reg1.fit(X, y)
```

```python
#print ('Coefficients: ', reg1.coef_)
#print ('Intercept: ', reg1.intercept_)
```

```python
plt.scatter(X, y, color = "red")
plt.plot(X, reg1.predict(X), color = "blue")
plt.show()
```

```python
plt.scatter(X, y, color = "red")
X_test = np.arange(1.0, 10.0, 0.01)
plt.plot(X_test, reg1.predict(np.array(X_test).reshape(-1, 1)), color = "blue")
plt.show()
```

```python
reg1.predict(np.array([11]).reshape(-1, 1))
```

# Random forest regression
