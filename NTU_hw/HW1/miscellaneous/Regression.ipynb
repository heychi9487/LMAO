# Simple linear regression

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
%matplotlib inline

dataset = pd.read_csv("slr-salary.csv")
#X = dataset["lv"]
X = np.array(dataset["lv"]).reshape(-1, 1)
y = dataset["salary"]
#print(X)
#print(y)

reg1 = LinearRegression()

reg1.fit(X, y)
```

```python
reg1.predict(np.array([11]).reshape(-1, 1))
```

```python
print ('Coefficients: ', reg1.coef_)
print ('Intercept: ', reg1.intercept_)
```

```python
plt.scatter(X, y, color = "red")
plt.plot(X, reg1.predict(X), color = "blue")
plt.show()
```

# Polynomial regression

```python
from sklearn.preprocessing import PolynomialFeatures

dataset = pd.read_csv("slr-salary.csv")
X = np.array(dataset["lv"]).reshape(-1, 1)
y = dataset["salary"]

poly_reg = PolynomialFeatures(degree = 2)

X_poly = poly_reg.fit_transform(X)
reg2 = LinearRegression()

reg2.fit(X_poly, y)
```

```python
print(reg2.predict(poly_reg.fit_transform([[11]])))
```

```python
print ('Coefficients: ', reg2.coef_)
print ('Intercept: ', reg2.intercept_)
```

```python
plt.scatter(X, y, color = "red")
plt.plot(X, reg2.predict(X_poly), color = "blue")
plt.show()
```

```python
poly_reg = PolynomialFeatures(degree = 6)

X_poly = poly_reg.fit_transform(X)
reg3 = LinearRegression()

reg3.fit(X_poly, y)

plt.scatter(X, y, color = "red")
plt.plot(X, reg3.predict(X_poly), color = "blue")
plt.show()
```

```python
print(reg3.predict(poly_reg.fit_transform([[11]])))
```

```python
print ('Coefficients: ', reg3.coef_)
print ('Intercept: ', reg3.intercept_)
```

# Multiple linear regression

```python
df = pd.read_csv("mlr-car.csv")

X = df[['Weight', 'Volume']]
y = df['CO2']

reg4 = LinearRegression()
reg4.fit(X, y)

#predict the CO2 emission of a car where the weight is 450kg, and the volume is 1500cm3:
predictedCO2 = reg4.predict([[450, 1500]])

print(predictedCO2)
```

```python
print ('Coefficients: ', reg4.coef_)
print ('Intercept: ', reg4.intercept_)
```

```python
predictedCO2withX = reg4.predict(X)
print(predictedCO2withX)
```
